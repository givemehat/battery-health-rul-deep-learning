## 1. Load and Inspect Processed NASA Battery Dataset

```python
df_nasa = pd.read_csv("data/processed/nasa_battery_cycles.csv")
print(f"Total cycle records: {len(df_nasa)}")
print(f"Total unique cells: {df_nasa['cell_id'].nunique()}")
df_nasa.head()
```

## 2. Visualize Capacity Degradation and Regeneration Dynamics

```python
plt.figure(figsize=(10, 6))
for cell in ["B0005", "B0006", "B0007", "B0018"]:
    data = df_nasa[df_nasa["cell_id"] == cell]
    plt.plot(data["cycle_index"], data["capacity"], label=f"{cell} (EOL: {EOL_THRESHOLDS.get(cell, 1.4)} Ah)", lw=2)
plt.axhline(1.4, color="red", linestyle=":", label="EOL 1.4 Ah (70% SOH)")
plt.axhline(1.5, color="green", linestyle=":", label="B0007 EOL 1.5 Ah")
plt.xlabel("Cycle Index", fontweight="bold")
plt.ylabel("Discharge Capacity (Ah)", fontweight="bold")
plt.title("Discharge Capacity Degradation Curves", fontweight="bold")
plt.legend()
plt.show()
```
